# 2. Weekly graph question

Below are the graphs from last week's weekly graph question.  Suppose you wanted to cause the viewer to examine the last three months of data first.  How could you use preattentive attributes to do that?  Suggest several possibilities.

```python
import numpy as np
import seaborn as sns
import pandas as pd
```

```python
numdates = 12
np.random.seed(0)
time_series = 100 * (np.random.normal(size = numdates) / 5 + np.arange(numdates) / 16)
datearray = pd.date_range(start='2023/01/01', end='2024/01/01', freq='M')
ax = sns.lineplot(x = datearray, y = time_series)
ax.set_xlabel("Year and month")
ax.set_ylabel("Nitrate level (% of eutrophic value)")
```

```python
np.random.seed(1)
time_series = 100 * (1 - (np.random.normal(size = numdates) / 8 + np.arange(numdates) / 16))
datearray = pd.date_range(start='2023/01/01', end='2024/01/01', freq='M')
ax = sns.lineplot(x = datearray, y = time_series)
ax.set_xlabel("Year and month")
ax.set_ylabel("Phosphate level (% of eutrophic value)")
```

#### Problem 2 Written Answer: 

Suggestions to use preattentive attributes to cause viewer to examine last three months first. 

Color: 
    - Make the last three months a brighter and bolder color while keeping the earlier colors more tame or neutral. 

Size: 
    - Increase the line width or dot size for the last three months. 

Enclosure: 
    - Add a highlight box to section off the last three months

Annotation: 
    - Add a label such as "Latest data" to signal the last three months

Reproduce any graph of your choice in p. 99-119 of the Storytelling With Data book as best you can. (The first half of chapter four). You do not have to get the exact data values right, just the overall look and feel.

```python
#recreation of figure 4.12 from pg 114
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 
          'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

processed = [150, 180, 260, 190, 170, 180, 150, 170, 160, 140, 150, 165]
received =  [160, 190, 250, 200, 180, 175, 160, 220, 190, 170, 180, 200]

plt.figure(figsize=(8, 5))

plt.plot(months, received, color='gray', linewidth=3, label='Received')

plt.plot(months, processed, color='#1f4e79', linewidth=3, label='Processed')

plt.ylabel('Count')
plt.xticks(months)
plt.legend()

plt.box(False)
plt.grid(False)

plt.show()
```
