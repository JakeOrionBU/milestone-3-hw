# 2. Weekly graph question

Suppose you plot the number visitors to a museum over the seven days of a week; then again the following week.  Can you suggest the advantages and disadvantages of the following two ways of presenting the data?

```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(0)
time_series_0 = np.array([100, 110, 130, 120, 100, 90, 80])
time_series_1 = np.array([70, 90, 100, 100, 120, 110, 90])
fig, axs = plt.subplots(1, 2)
axs[0].plot(time_series_0)
axs[1].plot(time_series_1)
plt.subplots_adjust(wspace = 0)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

```python
plt.plot(time_series_0)
plt.plot(time_series_1)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

### Answer for Question 2

When comparing two weeks of museum visitor data, side-by-side plots and overlaid line plots each have their strengths and weaknesses. Side-by-side plots offer a clear and uncluttered view of each week's trend, making it easier for viewers to interpret patterns without confusion, especially if they’re unfamiliar with multi-line charts. However, this format makes it harder to directly compare specific values between weeks, since it requires looking back and forth between the plots. In contrast, overlaid line plots allow for immediate comparison between weeks within the same visual frame, making it easy to spot differences, shifts, or consistencies in visitor patterns. This approach is more space-efficient and effective for highlighting trends, but can become cluttered and harder to interpret if not well-labeled, especially when more than two lines are shown. Overall, overlaid plots are best for direct comparison, while side-by-side plots are ideal for simpler, standalone insights.
